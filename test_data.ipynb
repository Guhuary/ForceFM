```python
import torch
epoch = 1
ckpt = f'workdir/base_finetune_fg_fd/epoch_{epoch:03d}.ckpt'
```

```python
state_dict = torch.load(ckpt, map_location='cpu', weights_only=False)
```

```python
state_dict.keys()
```

```python
sd = state_dict['state_dict']
```

```python
sd.keys()
```

```python
from collections import OrderedDict
new_dict = OrderedDict()
for k, v in sd.items():
	if 'model.' in k:
		new_dict[k.replace('model.', '')] = v
new_dict.keys()
```

```python
from omegaconf import OmegaConf
conf = OmegaConf.load('workdir/basemodel/hparams.yaml')
# fm_model_args = conf.args
fm_model_args = OmegaConf.merge(conf.args, conf.data.args)
```

```python
from src.models.get_model import get_vector_field
```

```python
model = get_vector_field(fm_model_args)
model.load_state_dict(new_dict)
```

```python
sd_n = model.state_dict()
torch.save(sd_n, f'workdir/base_finetune_fg_fd/epoch{epoch}_ema.ckpt')
```
